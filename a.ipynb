```python
import os
from fabric import Connection

from slurm_executor.pipeline.RSyncWorkspace import compose_rsync_command
from dotenv import load_dotenv

load_dotenv()

remote = os.getenv("SLURM_REMOTE")
port = os.getenv("SLURM_PORT")
user = os.getenv("SLURM_USERNAME")

command = compose_rsync_command(
    port=port,
    user=user,
    host=remote,
    local_root="./",
    remote_root=f"/home/{user}/remote_job/",
    inclusion_file="rsync-include.txt",
    direction="from_remote",
    dry_run=True,
)
with Connection(port=port, user=user, host=remote) as conn:
    result = conn.local(command, pty=False)
    print("stdout")
    print(result.stdout)

import jc

data = jc.parse("rsync", result.stdout)
print(data)
```
